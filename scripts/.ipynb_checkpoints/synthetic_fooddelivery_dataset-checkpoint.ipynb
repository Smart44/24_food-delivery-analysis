```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
# 1. Կարդում ենք ֆայլը
df = pd.read_csv('../data/synthetic_fooddelivery_dataset.csv')
```

```python
# 2. Նայում ենք առաջին 5 տողերին
print(df.head())
```

```python
# 3. Նայում ենք սյունակների անուններին և տիպերին
print(df.info())
```

```python
# 1. Ժամանակը դարձնում ենք datetime
df['Waktu_Transaksi'] = pd.to_datetime(df['Waktu_Transaksi'])
```

```python
# 2. Հեռավորության (KM) դատարկ տեղերը լրացնում ենք միջին արժեքով (median)
df['Jarak_Kirim_KM'] = df['Jarak_Kirim_KM'].fillna(df['Jarak_Kirim_KM'].median())
```

```python
# 3. Ռեյտինգի բացակայող տեղերը լրացնում ենք 0-ով կամ "No Rating"
df['Rating_Pelanggan'] = df['Rating_Pelanggan'].fillna(0)
```

```python
print(df.isnull().sum()) # Ստուգում ենք՝ արդյոք դեռ դատարկ տեղեր մնացին
```

```python
# 1. Ժամանակը դարձնում ենք datetime (ավելացնում ենք errors='coerce', որպեսզի սխալ տողերը չկանգնեցնեն պրոցեսը)
df['Waktu_Transaksi'] = pd.to_datetime(df['Waktu_Transaksi'], errors='coerce')
```

```python
# 2. Ստուգում ենք, թե արդյոք NaT (Not a Time) առաջացավ
print(f"Անվավեր ամսաթվեր: {df['Waktu_Transaksi'].isnull().sum()}")
```

```python
# Այժմ փորձիր այս գրաֆիկը
plt.figure(figsize=(10,6))
sns.scatterplot(data=df, x='Jarak_Kirim_KM', y='Waktu_Tunggu_Menit', alpha=0.5)
plt.title('Distance vs Waiting Time')
plt.xlabel('Distance (KM)')
plt.ylabel('Waiting Time (Minutes)')
plt.grid(True)
plt.savefig('../outputs/Distance vs Waiting Time.png')
plt.show()
```

```python
plt.figure(figsize=(12,6))
sns.boxplot(data=df, x='Kategori_Menu', y='Waktu_Tunggu_Menit')
plt.xticks(rotation=45)
plt.title('Waiting Time by Menu Category')
plt.savefig('../outputs/Waiting Time by Menu Category.png')
plt.show()
```

```python
# Փորձում ենք հասկանալ՝ ինչ տեսք ունեն այդ սխալ տողերը
print("Սխալ տվյալների օրինակներ:")
print(df[df['Waktu_Transaksi'].isnull()]['Waktu_Transaksi'].head())

# Եթե տվյալները շատ վատն են, կարող ենք ուղղակի ջնջել այդ տողերը, 
# որպեսզի մաքուր վերլուծություն անենք ժամերի հետ կապված
#df_clean = df.dropna(subset=['Waktu_Transaksi'])

# Ավելացնում ենք «Պատվերի ժամը» սյունակը
#df_clean['Hour'] = df_clean['Waktu_Transaksi'].dt.hour


# Ճիշտ ձևը՝ սկզբում սարքել լիարժեք պատճեն
df_clean = df.dropna(subset=['Waktu_Transaksi']).copy()
df_clean['Hour'] = df_clean['Waktu_Transaksi'].dt.hour


# Տեսնենք պատվերների քանակը ըստ ժամերի
plt.figure(figsize=(10,6))
sns.countplot(data=df_clean, x='Hour')
plt.title('Orders by Hour of the Day')
plt.savefig('../outputs/Orders by Hour of the Day.png')
plt.show()
```
